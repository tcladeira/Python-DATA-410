```python
import csv
```

```python
makes = ["Toyota", "Honda", "Ford", "Chevrolet"]
models = ["Camry", "Civic", "F-150", "Impala"]
max_speeds = [120, 130, 110, 115]
powers = [200, 180, 250, 220]
```

```python
car_list = list(zip(makes, models, max_speeds, powers))

car_list
```

```python
with open('../dataset/Cars.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['Make', 'Model', 'Max Speed', 'Power'])
    writer.writerows(car_list)

with open('../dataset/Cars.csv') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```
